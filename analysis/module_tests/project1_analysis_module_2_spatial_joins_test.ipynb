## Question 1: Join and Reconcile (15 points)

Complete a left point-in-polygon join, report unmatched points, and produce one row for every zone including zero-count zones.

```python
import geopandas as gpd
from shapely.geometry import Point, box

zones = gpd.GeoDataFrame(
    {"zone_id": ["A", "B", "C"]},
    geometry=[box(0, 0, 5, 5), box(5, 0, 10, 5), box(10, 0, 15, 5)],
    crs="EPSG:3347",
)
points = gpd.GeoDataFrame(
    {"point_id": [1, 2, 3, 4]},
    geometry=[Point(1, 1), Point(2, 2), Point(7, 2), Point(20, 20)],
    crs=zones.crs,
)

joined = gpd.GeoDataFrame(columns=["point_id", "zone_id", "geometry"])
unmatched = []
summary = gpd.GeoDataFrame(columns=["zone_id", "point_count", "geometry"])
# Complete the join, reconciliation, aggregation, and zero fill.

assert len(joined) == 4
assert unmatched == [4]
assert dict(zip(summary["zone_id"], summary["point_count"])) == {
    "A": 2,
    "B": 1,
    "C": 0,
}
```
