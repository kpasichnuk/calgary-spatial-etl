# Part 3: Copies and Mutation

Function parameters initially refer to the same objects supplied by the caller. Copying a DataFrame before column assignment protects the caller's object.

## Exercise 3A

Complete the function so it adds `reviewed` only to the returned DataFrame.

````python
def add_review_column(frame: pd.DataFrame) -> pd.DataFrame:
    # TODO: copy frame, add a reviewed column filled with None, and return the copy.
    frame = frame.copy()
    frame["reviewed"] = None
    return frame
````

````python
original_3a = pd.DataFrame({"site_id": [1, 2]})
result_3a = add_review_column(original_3a)

assert "reviewed" not in original_3a.columns, "The function changed its input. Did you copy first?"
assert "reviewed" in result_3a.columns
assert result_3a["reviewed"].isna().all()
assert result_3a is not original_3a
passed("Exercise 3A")
````

# Part 7: Combine Skills in Smaller Helpers

These exercises separate Stage 4's responsibilities into two functions. First ensure columns exist; then select the delivery schema.

## Exercise 7A: Ensure Columns

Copy the input, create absent required fields with `None`, and return the copy. Keep extra columns for now.

````python
def ensure_columns(
    frame: pd.DataFrame, required_fields: list[str]
) -> pd.DataFrame:
    # TODO: copy, loop, conditionally create columns, and return.
    frame = frame.copy()
    for field in required_fields:
        if field not in frame:
            frame[field] = None
    return frame
````

````python
original_7a = pd.DataFrame({"name": ["A"], "extra": [99]})
result_7a = ensure_columns(original_7a, ["name", "sector"])

assert list(original_7a.columns) == ["name", "extra"]
assert list(result_7a.columns) == ["name", "extra", "sector"]
assert result_7a["sector"].isna().all()
assert result_7a.loc[0, "name"] == "A"
passed("Exercise 7A")
````

## Exercise 7B: Select a Schema

Assume all required fields already exist. Keep them in the requested order, add and place `geometry` last, and remove extras.

````python
def select_schema(
    frame: pd.DataFrame, required_fields: list[str]
) -> pd.DataFrame:
    # TODO: build the output column list and return that selection.
    output_columns = []
    for field in required_fields:
        output_columns.append(field)
    output_columns.append("geometry")
    return frame[output_columns]
````

````python
frame_7b = pd.DataFrame({
    "extra": [99],
    "sector": ["North"],
    "geometry": ["POINT (0 0)"],
    "name": ["A"],
})
result_7b = select_schema(frame_7b, ["name", "sector"])

assert list(result_7b.columns) == ["name", "sector", "geometry"]
assert "extra" not in result_7b.columns
passed("Exercise 7B")
````

## Exercise 7C: Compose the Helpers

Call `ensure_columns` first and `select_schema` second. Do not rewrite either helper.

````python
original_7c = pd.DataFrame({
    "name": ["A"],
    "extra": [99],
    "geometry": ["POINT (0 0)"],
})
required_7c = ["name", "sector", "population"]

# TODO: call ensure_columns, then select_schema.
with_required_7c = ensure_columns(original_7c, required_7c)
result_7c = select_schema(with_required_7c, required_7c)
````

````python
assert list(result_7c.columns) == ["name", "sector", "population", "geometry"]
assert "extra" not in result_7c.columns
assert result_7c[["sector", "population"]].isna().all().all()
assert list(original_7c.columns) == ["name", "extra", "geometry"]
passed("Exercise 7C")
````

## Capstone A: Prepare an Inspection Table

Write a function that:

- Does not change the caller's DataFrame.
- Ensures every requested field exists.
- Fills absent fields with `None`.
- Records absent fields in requested order.
- Keeps only requested fields followed by `geometry`.
- Returns the prepared table and missing-field list.

````python
def prepare_inspection_table(
    frame: pd.DataFrame, required_fields: list[str]
) -> tuple[pd.DataFrame, list[str]]:
    # Type your complete solution here.
    frame = frame.copy()
    absent_fields = []
    for field in required_fields:
        if field not in frame:
            absent_fields.append(field)
            frame[field] = None
    required_fields.append("geometry")
    return frame[required_fields], absent_fields
````

````python
original_capstone_a = pd.DataFrame({
    "asset_id": [101],
    "condition": ["good"],
    "internal_note": ["do not deliver"],
    "geometry": ["POINT (2 3)"],
})

prepared_a, absent_a = prepare_inspection_table(
    original_capstone_a,
    ["asset_id", "condition", "inspection_date", "inspector"],
)

assert list(prepared_a.columns) == [
    "asset_id", "condition", "inspection_date", "inspector", "geometry"
]
assert absent_a == ["inspection_date", "inspector"]
assert prepared_a[["inspection_date", "inspector"]].isna().all().all()
assert list(original_capstone_a.columns) == [
    "asset_id", "condition", "internal_note", "geometry"
]
passed("Capstone A")
````

## Capstone B: Prepare a GeoDataFrame

Write a new function from the same requirements, using a GeoDataFrame. The output must also preserve GeoDataFrame type, geometry, and CRS. Do not call the Stage 4 assessment function.

````python
def prepare_site_delivery(
    gdf: gpd.GeoDataFrame, required_fields: list[str]
) -> tuple[gpd.GeoDataFrame, list[str]]:
    # Type your complete solution here.
    missing_fields = []
    gdf = gdf.copy()

    for field in required_fields:
        if field not in gdf.columns:
            missing_fields.append(field)
            gdf[field] = None

    output_columns  = required_fields +  ["geometry"]

    return gdf[output_columns], missing_fields
````

````python
original_capstone_b = gpd.GeoDataFrame(
    {
        "site_id": [1, 2],
        "site_name": ["A", "B"],
        "internal_notes": ["x", "y"],
    },
    geometry=[Point(-114.0, 51.0), Point(-114.1, 51.1)],
    crs="EPSG:4326",
)
required_capstone_b = ["site_id", "site_name", "status", "review_date"]

prepared_b, absent_b = prepare_site_delivery(
    original_capstone_b, required_capstone_b
)

assert isinstance(prepared_b, gpd.GeoDataFrame)
assert list(prepared_b.columns) == required_capstone_b + ["geometry"]
assert absent_b == ["status", "review_date"]
assert prepared_b.crs == original_capstone_b.crs
assert prepared_b.geometry.equals(original_capstone_b.geometry)
assert "status" not in original_capstone_b.columns
assert "internal_notes" not in prepared_b.columns
passed("Capstone B")
````
